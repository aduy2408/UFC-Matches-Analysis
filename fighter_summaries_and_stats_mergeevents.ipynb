```python
import numpy as np
import pandas as pd
from datetime import datetime
```

```python
df_ufc_event = pd.read_csv('ufc_event_details_processed.csv')
df_fight_results = pd.read_csv('ufc_fight_results_processed_official.csv')
df_fight_stats = pd.read_csv('ufc_fight_stats_processed.csv')
```

## Creating summary files of names & fighters' history

```python
df_fight_results['BOUT'] = df_fight_results['BOUT'].str.lower().str.strip()
df_fight_results['BOUT'] = df_fight_results['BOUT'].str.replace("  vs.", " vs.", regex=False)
df_fight_stats['BOUT'] = df_fight_stats['BOUT'].str.lower().str.strip()
df_fight_results['EVENT'] = df_fight_results['EVENT'].str.lower().str.strip()
df_fight_stats['EVENT'] = df_fight_stats['EVENT'].str.lower().str.strip()
df_ufc_event['EVENT'] = df_ufc_event['EVENT'].str.lower().str.strip()  
```

```python
def clean_bout_name(bout_name):
    return " vs. ".join(sorted(bout_name.lower().strip().replace("  ", " ").split(" vs. ")))
df_fight_stats["BOUT"] = df_fight_stats["BOUT"].apply(clean_bout_name)
df_fight_results["BOUT"] = df_fight_results["BOUT"].apply(clean_bout_name)
```

## Adding weightclass to the fight_stats for easier processing

```python
fight_stats_with_weightclass = df_fight_stats.merge(

    df_fight_results[['EVENT', 'BOUT', 'weightclass_numeric','weight_class','gender','is_title_bout']],

    on=['EVENT', 'BOUT'],

    how='left'
)
```

```python
missing_values_after_merge = fight_stats_with_weightclass[['weightclass_numeric', 'weight_class', 'gender', 'is_title_bout']].isnull().sum()
```

```python
fight_stats_with_weightclass = fight_stats_with_weightclass.merge(

    df_ufc_event[['EVENT','DATE', 'LOCATION']],

    on=['EVENT'],

    how='left'
)
```

```python
fight_results_with_locale = df_fight_results.merge(
    df_ufc_event[['EVENT','DATE',"LOCATION"]],
    on = ['EVENT'],
    how = 'left'
)
```

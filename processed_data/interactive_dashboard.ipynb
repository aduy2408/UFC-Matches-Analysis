```python
    @app.callback(
        [Output("timeline-slider", "max"),
        Output("timeline-slider", "value"),
        Output("timeline-slider", "marks")],
        Input("match-dropdown", "value")
    )
    def update_timeline_slider_props(match_name):
        if not match_name or match_name not in stats_df['BOUT'].unique():
            max_round = 3
        else:
            bout_stats = stats_df[stats_df['BOUT'] == match_name]
            max_round = int(bout_stats['ROUND'].max()) if not bout_stats.empty else 3
        
        # Create marks for the slider based on max_round
        marks = {i: {'label': f'R{i}', 'style': {'color': 'white'}} for i in range(1, max_round + 1)}
        
        return max_round, max_round, marks
    
    @app.callback(
        Output("fight-timeline", "figure"),
        [Input("match-dropdown", "value"),
        Input("timeline-slider", "value")],
    )
    def update_fight_timeline(match_name, selected_round):
        if not match_name:
            return {}
            
        # Get fight stats for the selected match
        bout_stats = stats_df[stats_df['BOUT'] == match_name].copy()
        if bout_stats.empty:
            return {}
            
        fighters = bout_stats['FIGHTER'].unique()
        if len(fighters) != 2:
            return {}
        
        # Ensure ROUND is an integer
        bout_stats['ROUND'] = pd.to_numeric(bout_stats['ROUND'], errors='coerce')
        bout_stats = bout_stats.dropna(subset=['ROUND'])
        bout_stats['ROUND'] = bout_stats['ROUND'].astype(int)
        
        # Convert selected_round to int if it's not already
        if selected_round is not None:
            selected_round = int(selected_round)
        else:
            selected_round = bout_stats['ROUND'].max()
        
        # Create subplot with line chart and two pie charts
        fig = make_subplots(
            rows=2, cols=2,
            specs=[
                [{"colspan": 2}, None],
                [{"type": "pie"}, {"type": "pie"}]
            ],
            subplot_titles=(
                f"Cumulative Stats Through Round {selected_round}",
                f"{fighters[0]} - Round {selected_round} Strikes",
                f"{fighters[1]} - Round {selected_round} Strikes"
            ),
            vertical_spacing=0.15
        )
        
        # Filter stats up to the selected round for line chart
        filtered_stats = bout_stats[bout_stats['ROUND'] <= selected_round]
        
        # Prepare data for each fighter
        for fighter_idx, fighter in enumerate(fighters):
            fighter_stats = filtered_stats[filtered_stats['FIGHTER'] == fighter]
            
            if fighter_stats.empty:
                continue
                
            # Group by round for cumulative stats
            round_stats = fighter_stats.groupby('ROUND').agg({
                'sig_str_land': 'sum',
                'sig_str_attempt': 'sum',
                'total_str_land': 'sum',
                'takedown_land': 'sum',
                'touchdown_attempt': 'sum'
            }).reset_index()
            
            # Calculate cumulative stats
            round_stats['cum_sig_strikes'] = round_stats['sig_str_land'].cumsum()
            round_stats['cum_takedowns'] = round_stats['takedown_land'].cumsum()
            
            # Add trace for significant strikes
            fig.add_trace(
                go.Scatter(
                    x=round_stats['ROUND'],
                    y=round_stats['cum_sig_strikes'],
                    mode='lines+markers',
                    name=f"{fighter} - Sig. Strikes",
                    line=dict(width=3, color='red' if fighter_idx == 0 else 'blue'),
                    marker=dict(size=8)
                ),
                row=1, col=1
            )
            
            # Add trace for takedowns
            fig.add_trace(
                go.Scatter(
                    x=round_stats['ROUND'],
                    y=round_stats['cum_takedowns'] * 5,  # Scale for visibility
                    mode='lines+markers',
                    name=f"{fighter} - Takedowns (×5)",
                    line=dict(width=2, dash='dot', color='darkred' if fighter_idx == 0 else 'darkblue'),
                    marker=dict(symbol='diamond', size=8)
                ),
                row=1, col=1
            )
            
            # Get data for the selected round only (for pie charts)
            selected_round_stats = fighter_stats[fighter_stats['ROUND'] == selected_round]
            
            if not selected_round_stats.empty:
                # Get strike data for selected round
                head_strikes = selected_round_stats['total_head_land'].sum()
                body_strikes = selected_round_stats['total_body_land'].sum()
                leg_strikes = selected_round_stats['total_leg_land'].sum()
                
                # Add pie chart for strike distribution in the selected round
                colors = ["#636EFA", "#EF553B", "#00CC96"]  # Blue, Red, Green
                
                fig.add_trace(
                    go.Pie(
                        labels=['Head', 'Body', 'Leg'],
                        values=[head_strikes, body_strikes, leg_strikes],
                        name=f"{fighter} - Round {selected_round}",
                        marker=dict(colors=colors),
                        textinfo='percent',
                        hoverinfo='label+value+percent',
                        hole=0.3,
                    ),
                    row=2, col=fighter_idx+1
                )
        
        # Update layout
        fig.update_layout(
            title=f"Fight Progression: {match_name}",
            height=700,  # Increased to accommodate pie charts
            paper_bgcolor="rgba(20, 20, 20, 1)",
            plot_bgcolor="rgba(30, 30, 30, 1)",
            font=dict(color="white"),
            legend=dict(
                orientation="h",
                font=dict(color="white"),
                y=0.5
            ),
        )
        
        # Update x-axis for the line chart
        fig.update_xaxes(
            title="Round",
            tickmode='array',
            tickvals=list(range(1, int(bout_stats['ROUND'].max()) + 1)),
            ticktext=[f"R{r}" for r in range(1, int(bout_stats['ROUND'].max()) + 1)],
            row=1, col=1
        )
        
        # Update y-axis for the line chart
        fig.update_yaxes(
            title="Cumulative Stats",
            row=1, col=1
        )
        
        return fig
```

```python
import dash
from dash import dcc, html, callback, Input, Output
import dash_bootstrap_components as dbc
import pandas as pd
import plotly.express as px
import plotly.graph_objects as go
import numpy as np

# Initialize the Dash app with Bootstrap theme
app = dash.Dash(__name__, external_stylesheets=[dbc.themes.CYBORG], use_pages=False)

# Sample data - in a real app, you would connect to an API or database
# Fighters data
fighters_data = pd.DataFrame({
    'name': ['Jon Jones', 'Khabib Nurmagomedov', 'Israel Adesanya', 'Francis Ngannou', 
             'Amanda Nunes', 'Valentina Shevchenko', 'Dustin Poirier', 'Charles Oliveira'],
    'weight_class': ['Heavyweight', 'Lightweight', 'Middleweight', 'Heavyweight', 
                     'Bantamweight', 'Flyweight', 'Lightweight', 'Lightweight'],
    'record': ['27-1-0', '29-0-0', '24-2-0', '17-3-0', '22-5-0', '23-4-0', '29-7-0', '33-9-0'],
    'knockouts': [10, 8, 16, 12, 13, 8, 14, 9],
    'submissions': [7, 11, 0, 4, 4, 7, 8, 16],
    'striking_accuracy': [0.68, 0.48, 0.74, 0.53, 0.62, 0.66, 0.53, 0.51],
    'takedown_accuracy': [0.45, 0.70, 0.30, 0.25, 0.40, 0.55, 0.35, 0.42],
    'win_streak': [5, 13, 3, 6, 2, 7, 0, 1]
})

# Matches data
matches_data = pd.DataFrame({
    'match_id': range(1, 11),
    'fighter1': ['Jon Jones', 'Khabib Nurmagomedov', 'Israel Adesanya', 'Francis Ngannou', 
                 'Jon Jones', 'Khabib Nurmagomedov', 'Israel Adesanya', 'Amanda Nunes', 
                 'Valentina Shevchenko', 'Charles Oliveira'],
    'fighter2': ['Ciryl Gane', 'Dustin Poirier', 'Alex Pereira', 'Stipe Miocic', 
                 'Dominick Reyes', 'Conor McGregor', 'Robert Whittaker', 'Julianna Peña',
                 'Amanda Nunes', 'Dustin Poirier'],
    'event': ['UFC 285', 'UFC 242', 'UFC 287', 'UFC 260', 
              'UFC 247', 'UFC 229', 'UFC 271', 'UFC 269',
              'UFC 277', 'UFC 269'],
    'date': ['2023-03-04', '2019-09-07', '2023-04-08', '2021-03-27',
             '2020-02-08', '2018-10-06', '2022-02-12', '2021-12-11',
             '2022-07-30', '2021-12-11'],
    'winner': ['Jon Jones', 'Khabib Nurmagomedov', 'Israel Adesanya', 'Francis Ngannou',
              'Jon Jones', 'Khabib Nurmagomedov', 'Israel Adesanya', 'Julianna Peña',
              'Amanda Nunes', 'Charles Oliveira'],
    'method': ['Submission', 'Submission', 'KO', 'KO', 
               'Decision', 'Submission', 'Decision', 'Submission',
               'Decision', 'Submission'],
    'round': [1, 3, 2, 2, 5, 4, 5, 2, 5, 3],
    'time': ['2:04', '3:15', '4:21', '0:52', '5:00', '3:03', '5:00', '3:26', '5:00', '1:40']
})

# Create the navbar with UFC branding
navbar = dbc.Navbar(
    dbc.Container(
        [
            html.A(
                dbc.Row(
                    [
                        dbc.Col(html.Img(src="https://upload.wikimedia.org/wikipedia/commons/thumb/9/92/UFC_Logo.svg/2560px-UFC_Logo.svg.png", height="50px")),
                        dbc.Col(html.H3("UFC ANALYTICS", className="ms-3 text-light")),
                    ],
                    align="center",
                ),
                style={"textDecoration": "none"},
            ),
            dbc.Nav(
                [
                    dbc.NavItem(dbc.NavLink("FIGHTER STATS", href="#fighters", id="fighters-link")),
                    dbc.NavItem(dbc.NavLink("MATCH ANALYSIS", href="#matches", id="matches-link")),
                    dbc.NavItem(dbc.NavLink("FIGHTER COMPARISON", href="#comparison", id="comparison-link")),
                ],
                className="ms-auto",
                navbar=True,
            ),
        ]
    ),
    color="dark",
    dark=True,
    className="mb-4",
)

# Create the tabs content
# 1. Fighter Stats Tab
fighter_stats_tab = dbc.Container([
    dbc.Row([
        dbc.Col([
            html.H4("SELECT FIGHTER", className="text-danger"),
            dcc.Dropdown(
                id='fighter-dropdown',
                options=[{'label': name, 'value': name} for name in fighters_data['name']],
                value=fighters_data['name'][0],
                className="mb-4"
            ),
        ], width=4),
        dbc.Col([
            html.H4("FILTER OPTIONS", className="text-danger"),
            dbc.Row([
                dbc.Col([
                    html.Label("Weight Class"),
                    dcc.Dropdown(
                        id='weight-class-filter',
                        options=[{'label': wc, 'value': wc} 
                                 for wc in sorted(fighters_data['weight_class'].unique())],
                        multi=True,
                        placeholder="All Weight Classes"
                    ),
                ], width=6),
                dbc.Col([
                    html.Label("Min Win Streak"),
                    dcc.Slider(
                        id='win-streak-slider',
                        min=0,
                        max=fighters_data['win_streak'].max(),
                        step=1,
                        value=0,
                        marks={i: str(i) for i in range(0, fighters_data['win_streak'].max() + 1, 2)}
                    ),
                ], width=6),
            ]),
        ], width=8),
    ], className="mb-4"),
    
    dbc.Row([
        dbc.Col([
            dbc.Card([
                dbc.CardHeader(html.H5("FIGHTER PROFILE", className="text-center")),
                dbc.CardBody(
                    html.Div(id="fighter-profile-content")
                )
            ]),
        ], width=4),
        dbc.Col([
            dbc.Card([
                dbc.CardHeader(html.H5("PERFORMANCE METRICS", className="text-center")),
                dbc.CardBody(
                    dcc.Graph(id="fighter-radar-chart", style={"height": "400px"})
                )
            ]),
        ], width=8),
    ], className="mb-4"),
    
    dbc.Row([
        dbc.Col([
            dbc.Card([
                dbc.CardHeader(html.H5("RECENT FIGHTS", className="text-center")),
                dbc.CardBody(
                    html.Div(id="fighter-recent-fights")
                )
            ]),
        ]),
    ]),
])

# 2. Match Analysis Tab
match_analysis_tab = dbc.Container([
    dbc.Row([
        dbc.Col([
            html.H4("SELECT EVENT", className="text-danger"),
            dcc.Dropdown(
                id='event-dropdown',
                options=[{'label': event, 'value': event} for event in sorted(matches_data['event'].unique())],
                value=matches_data['event'][0],
                className="mb-4"
            ),
        ], width=4),
        dbc.Col([
            html.H4("FILTER OPTIONS", className="text-danger"),
            dbc.Row([
                dbc.Col([
                    html.Label("Result Type"),
                    dcc.Dropdown(
                        id='result-filter',
                        options=[{'label': method, 'value': method} 
                                 for method in sorted(matches_data['method'].unique())],
                        multi=True,
                        placeholder="All Results"
                    ),
                ], width=6),
                dbc.Col([
                    html.Label("Date Range"),
                    dcc.DatePickerRange(
                        id='date-range',
                        min_date_allowed=min(pd.to_datetime(matches_data['date'])),
                        max_date_allowed=max(pd.to_datetime(matches_data['date'])),
                        start_date=min(pd.to_datetime(matches_data['date'])),
                        end_date=max(pd.to_datetime(matches_data['date'])),
                    ),
                ], width=6),
            ]),
        ], width=8),
    ], className="mb-4"),
    
    dbc.Row([
        dbc.Col([
            dbc.Card([
                dbc.CardHeader(html.H5("MATCH DETAILS", className="text-center")),
                dbc.CardBody(
                    html.Div(id="match-details-content")
                )
            ]),
        ], width=5),
        dbc.Col([
            dbc.Card([
                dbc.CardHeader(html.H5("FIGHT STATISTICS", className="text-center")),
                dbc.CardBody(
                    dcc.Graph(id="match-stats-chart", style={"height": "400px"})
                )
            ]),
        ], width=7),
    ], className="mb-4"),
    
    dbc.Row([
        dbc.Col([
            dbc.Card([
                dbc.CardHeader(html.H5("FINISH DISTRIBUTION", className="text-center")),
                dbc.CardBody(
                    dcc.Graph(id="finish-distribution", style={"height": "300px"})
                )
            ]),
        ], width=6),
        dbc.Col([
            dbc.Card([
                dbc.CardHeader(html.H5("ROUND FINISHES", className="text-center")),
                dbc.CardBody(
                    dcc.Graph(id="round-finishes", style={"height": "300px"})
                )
            ]),
        ], width=6),
    ]),
])

# 3. Fighter Comparison Tab
fighter_comparison_tab = dbc.Container([
    dbc.Row([
        dbc.Col([
            html.H4("FIGHTER 1", className="text-danger"),
            dcc.Dropdown(
                id='fighter1-dropdown',
                options=[{'label': name, 'value': name} for name in fighters_data['name']],
                value=fighters_data['name'][0],
                className="mb-4"
            ),
        ], width=6),
        dbc.Col([
            html.H4("FIGHTER 2", className="text-danger"),
            dcc.Dropdown(
                id='fighter2-dropdown',
                options=[{'label': name, 'value': name} for name in fighters_data['name']],
                value=fighters_data['name'][1],
                className="mb-4"
            ),
        ], width=6),
    ], className="mb-4"),
    
    dbc.Row([
        dbc.Col([
            dbc.Card([
                dbc.CardHeader(html.H5("STATS COMPARISON", className="text-center")),
                dbc.CardBody(
                    dcc.Graph(id="fighters-comparison-chart", style={"height": "400px"})
                )
            ]),
        ]),
    ], className="mb-4"),
    
    dbc.Row([
        dbc.Col([
            dbc.Card([
                dbc.CardHeader(html.H5("HEAD-TO-HEAD", className="text-center")),
                dbc.CardBody(
                    html.Div(id="head-to-head-content")
                )
            ]),
        ], width=4),
        dbc.Col([
            dbc.Card([
                dbc.CardHeader(html.H5("PERFORMANCE HISTORY", className="text-center")),
                dbc.CardBody(
                    dcc.Graph(id="performance-history", style={"height": "300px"})
                )
            ]),
        ], width=8),
    ]),
])

# Create the content div that will hold the active tab
content = html.Div(
    fighter_stats_tab,  # Default to fighter stats tab
    id="page-content"
)

# Combine all sections
app.layout = html.Div([
    navbar,
    content,
    
    # Store the active tab
    dcc.Store(id='active-tab', data='fighters'),
])

# Callback to update the active tab
@app.callback(
    Output("page-content", "children"),
    [
        Input("fighters-link", "n_clicks"),
        Input("matches-link", "n_clicks"),
        Input("comparison-link", "n_clicks"),
        Input('active-tab', 'data')
    ]
)
def toggle_active_tab(fighters_clicks, matches_clicks, comparison_clicks, active_tab):
    ctx = dash.callback_context
    
    if not ctx.triggered:
        return fighter_stats_tab
        
    trigger_id = ctx.triggered[0]['prop_id'].split('.')[0]
    
    if trigger_id == "fighters-link":
        return fighter_stats_tab
    elif trigger_id == "matches-link":
        return match_analysis_tab
    elif trigger_id == "comparison-link":
        return fighter_comparison_tab
    else:
        # Handle tab change based on active-tab data
        if active_tab == 'fighters':
            return fighter_stats_tab
        elif active_tab == 'matches':
            return match_analysis_tab
        elif active_tab == 'comparison':
            return fighter_comparison_tab
        else:
            return fighter_stats_tab

# Callback to store active tab
@app.callback(
    Output('active-tab', 'data'),
    [
        Input("fighters-link", "n_clicks"),
        Input("matches-link", "n_clicks"),
        Input("comparison-link", "n_clicks")
    ]
)
def store_active_tab(fighters_clicks, matches_clicks, comparison_clicks):
    ctx = dash.callback_context
    
    if not ctx.triggered:
        return 'fighters'
        
    trigger_id = ctx.triggered[0]['prop_id'].split('.')[0]
    
    if trigger_id == "fighters-link":
        return 'fighters'
    elif trigger_id == "matches-link":
        return 'matches'
    elif trigger_id == "comparison-link":
        return 'comparison'
    else:
        return 'fighters'

# Fighter profile content callback
@app.callback(
    Output("fighter-profile-content", "children"),
    Input("fighter-dropdown", "value")
)
def update_fighter_profile(fighter_name):
    if not fighter_name:
        return html.Div("Please select a fighter")
    
    fighter = fighters_data[fighters_data['name'] == fighter_name].iloc[0]
    wins, losses, draws = map(int, fighter['record'].split('-'))
    
    return html.Div([
        html.Div([
            html.Img(src=f"assets/{fighter_name.lower().replace(' ', '_')}.jpg", 
                   style={'width': '150px', 'height': '150px', 'borderRadius': '50%'},
                   className="mx-auto d-block mb-3"),
            html.H3(fighter_name, className="text-center mb-3"),
            html.H5(f"Weight Class: {fighter['weight_class']}", className="text-center text-muted mb-2"),
            
            html.Div(className="d-flex justify-content-center mb-3", children=[
                html.Div(className="text-center mx-2", children=[
                    html.H2(wins, className="text-success m-0"),
                    html.P("WINS", className="text-muted m-0")
                ]),
                html.Div(className="text-center mx-2", children=[
                    html.H2(losses, className="text-danger m-0"),
                    html.P("LOSSES", className="text-muted m-0")
                ]),
                html.Div(className="text-center mx-2", children=[
                    html.H2(draws, className="text-warning m-0"),
                    html.P("DRAWS", className="text-muted m-0")
                ]),
            ]),
            
            html.H5("FINISH RATE", className="text-center mb-1"),
            html.Div(className="d-flex justify-content-center", children=[
                html.Div(className="text-center mx-2", children=[
                    html.H4(fighter['knockouts'], className="text-danger m-0"),
                    html.P("KO/TKO", className="text-muted m-0")
                ]),
                html.Div(className="text-center mx-2", children=[
                    html.H4(fighter['submissions'], className="text-primary m-0"),
                    html.P("SUBMISSIONS", className="text-muted m-0")
                ]),
            ]),
        ])
    ])

# Fighter radar chart callback
@app.callback(
    Output("fighter-radar-chart", "figure"),
    Input("fighter-dropdown", "value")
)
def update_radar_chart(fighter_name):
    if not fighter_name:
        return {}
    
    fighter = fighters_data[fighters_data['name'] == fighter_name].iloc[0]
    
    # Calculate stats for radar chart - normalizing values
    max_ko = fighters_data['knockouts'].max()
    max_sub = fighters_data['submissions'].max()
    max_streak = fighters_data['win_streak'].max()
    
    categories = ['Striking Accuracy', 'Takedown Accuracy', 'KO Power', 'Submission Skills', 'Win Streak']
    values = [
        fighter['striking_accuracy'],
        fighter['takedown_accuracy'],
        fighter['knockouts'] / max_ko if max_ko > 0 else 0,
        fighter['submissions'] / max_sub if max_sub > 0 else 0,
        fighter['win_streak'] / max_streak if max_streak > 0 else 0
    ]
    
    # Add the first value to close the loop
    categories = categories + [categories[0]]
    values = values + [values[0]]
    
    fig = go.Figure()
    
    fig.add_trace(go.Scatterpolar(
        r=values,
        theta=categories,
        fill='toself',
        name=fighter_name,
        line_color='red',
        fillcolor='rgba(255, 0, 0, 0.3)'
    ))
    
    fig.update_layout(
        polar=dict(
            radialaxis=dict(
                visible=True,
                range=[0, 1]
            )
        ),
        showlegend=False,
        margin=dict(l=60, r=60, t=40, b=40),
        paper_bgcolor='rgba(0, 0, 0, 0)',
        plot_bgcolor='rgba(0, 0, 0, 0)',
        font=dict(color='white')
    )
    
    return fig

# Fighter recent fights callback
@app.callback(
    Output("fighter-recent-fights", "children"),
    Input("fighter-dropdown", "value")
)
def update_recent_fights(fighter_name):
    if not fighter_name:
        return html.Div("Please select a fighter")
    
    # Get all fights where the fighter participated
    fighter_matches = matches_data[
        (matches_data['fighter1'] == fighter_name) | (matches_data['fighter2'] == fighter_name)
    ].sort_values('date', ascending=False)
    
    if fighter_matches.empty:
        return html.Div("No fight data available")
    
    return html.Div([
        dbc.Table([
            html.Thead([
                html.Tr([
                    html.Th("EVENT"),
                    html.Th("DATE"),
                    html.Th("OPPONENT"),
                    html.Th("RESULT"),
                    html.Th("METHOD"),
                    html.Th("ROUND"),
                ])
            ]),
            html.Tbody([
                html.Tr([
                    html.Td(row['event']),
                    html.Td(row['date']),
                    html.Td(row['fighter2'] if row['fighter1'] == fighter_name else row['fighter1']),
                    html.Td(html.Span("WIN", className="text-success") 
                           if row['winner'] == fighter_name else 
                           html.Span("LOSS", className="text-danger")),
                    html.Td(row['method']),
                    html.Td(f"R{row['round']} ({row['time']})")
                ]) for _, row in fighter_matches.iterrows()
            ])
        ], striped=True, bordered=True, hover=True, responsive=True)
    ])

# Match details callback
@app.callback(
    Output("match-details-content", "children"),
    Input("event-dropdown", "value")
)
def update_match_details(event_name):
    if not event_name:
        return html.Div("Please select an event")
    
    # Get all matches for the selected event
    event_matches = matches_data[matches_data['event'] == event_name]
    
    if event_matches.empty:
        return html.Div("No match data available")
    
    # For simplicity, take the first match from the event
    match = event_matches.iloc[0]
    
    return html.Div([
        html.H4(match['event'], className="text-center mb-3"),
        html.P(f"Date: {match['date']}", className="text-center text-muted mb-4"),
        
        html.Div(className="d-flex justify-content-between align-items-center mb-4", children=[
            html.Div(className="text-center", children=[
                html.Img(src=f"assets/{match['fighter1'].lower().replace(' ', '_')}.jpg", 
                       style={'width': '100px', 'height': '100px', 'borderRadius': '50%'}),
                html.H5(match['fighter1'], className="mt-2"),
                html.P(fighters_data[fighters_data['name'] == match['fighter1']].iloc[0]['weight_class'], 
                     className="text-muted")
            ]),
            
            html.Div(className="text-center", children=[
                html.H3("VS", className="text-danger")
            ]),
            
            html.Div(className="text-center", children=[
                html.Img(src=f"assets/{match['fighter2'].lower().replace(' ', '_')}.jpg", 
                       style={'width': '100px', 'height': '100px', 'borderRadius': '50%'}),
                html.H5(match['fighter2'], className="mt-2"),
                html.P(fighters_data[fighters_data['name'] == match['fighter2']].iloc[0]['weight_class'], 
                     className="text-muted")
            ]),
        ]),
        
        html.Div(className="text-center mb-3", children=[
            html.H5("RESULT"),
            html.H4([
                html.Span(f"Winner: {match['winner']}", 
                        className="text-success" if match['winner'] == match['fighter1'] else "text-danger"),
            ]),
            html.P(f"via {match['method']} - Round {match['round']} ({match['time']})")
        ])
    ])

# Match stats chart callback
@app.callback(
    Output("match-stats-chart", "figure"),
    Input("event-dropdown", "value")
)
def update_match_stats(event_name):
    if not event_name:
        return {}
    
    # In a real app, this would be actual match statistics
    # For this example, we'll generate random stats
    np.random.seed(42)  # For reproducibility
    
    event_matches = matches_data[matches_data['event'] == event_name]
    if event_matches.empty:
        return {}
    
    match = event_matches.iloc[0]
    fighter1 = match['fighter1']
    fighter2 = match['fighter2']
    
    # Sample fight statistics
    stats_categories = [
        'Significant Strikes', 'Total Strikes', 'Takedowns', 
        'Takedown Defense', 'Control Time (min)'
    ]
    
    # Generate sample stats
    fighter1_values = np.random.randint(10, 100, size=len(stats_categories))
    fighter2_values = np.random.randint(10, 100, size=len(stats_categories))
    
    # For control time, generate smaller numbers
    fighter1_values[-1] = np.random.randint(1, 15)
    fighter2_values[-1] = np.random.randint(1, 15)
    
    fig = go.Figure()
    
    fig.add_trace(go.Bar(
        x=stats_categories,
        y=fighter1_values,
        name=fighter1,
        marker_color='red'
    ))
    
    fig.add_trace(go.Bar(
        x=stats_categories,
        y=fighter2_values,
        name=fighter2,
        marker_color='blue'
    ))
    
    fig.update_layout(
        barmode='group',
        title='Fight Statistics',
        xaxis_title='',
        yaxis_title='Count',
        legend_title='Fighter',
        margin=dict(l=40, r=40, t=40, b=40),
        paper_bgcolor='rgba(0, 0, 0, 0)',
        plot_bgcolor='rgba(0, 0, 0, 0)',
        font=dict(color='white')
    )
    
    return fig

# Finish distribution callback
@app.callback(
    Output("finish-distribution", "figure"),
    Input("event-dropdown", "value")
)
def update_finish_distribution(_):
    # This doesn't depend on the specific event - showing overall UFC stats
    
    finish_types = ['KO/TKO', 'Submission', 'Decision', 'DQ/No Contest']
    finish_counts = [35, 28, 32, 5]  # Sample percentage distribution
    
    fig = go.Figure(data=[go.Pie(
        labels=finish_types,
        values=finish_counts,
        hole=.4,
        marker_colors=['#FF4136', '#0074D9', '#FFDC00', '#B10DC9']
    )])
    
    fig.update_layout(
        title='UFC Fight Outcomes',
        margin=dict(l=40, r=40, t=50, b=40),
        paper_bgcolor='rgba(0, 0, 0, 0)',
        plot_bgcolor='rgba(0, 0, 0, 0)',
        font=dict(color='white'),
        legend=dict(
            orientation="h",
            yanchor="bottom",
            y=0,
            xanchor="center",
            x=0.5
        )
    )
    
    return fig

# Round finishes callback
@app.callback(
    Output("round-finishes", "figure"),
    Input("event-dropdown", "value")
)
def update_round_finishes(_):
    # Sample data showing finishes by round
    rounds = ['Round 1', 'Round 2', 'Round 3', 'Round 4', 'Round 5']
    finishes = [42, 25, 18, 8, 7]  # Sample data
    
    fig = go.Figure()
    
    fig.add_trace(go.Bar(
        x=rounds,
        y=finishes,
        marker_color='#FF4136'
    ))
    
    fig.update_layout(
        title='Finishes by Round',
        xaxis_title='',
        yaxis_title='Number of Finishes',
        margin=dict(l=40, r=40, t=50, b=40),
        paper_bgcolor='rgba(0, 0, 0, 0)',
        plot_bgcolor='rgba(0, 0, 0, 0)',
        font=dict(color='white')
    )
    
    return fig

# Fighters comparison chart
@app.callback(
    Output("fighters-comparison-chart", "figure"),
    [Input("fighter1-dropdown", "value"),
     Input("fighter2-dropdown", "value")]
)
def update_fighters_comparison(fighter1_name, fighter2_name):
    if not fighter1_name or not fighter2_name:
        return {}
    
    fighter1 = fighters_data[fighters_data['name'] == fighter1_name].iloc[0]
    fighter2 = fighters_data[fighters_data['name'] == fighter2_name].iloc[0]
    
    categories = ['Striking Accuracy', 'Takedown Accuracy', 'KO Power', 'Submission Skills', 'Win Streak']
    
    # Scale values for comparison
    max_ko = fighters_data['knockouts'].max()
    max_sub = fighters_data['submissions'].max()
    max_streak = fighters_data['win_streak'].max()
    
    fighter1_values = [
        fighter1['striking_accuracy'],
        fighter1['takedown_accuracy'],
        fighter1['knockouts'] / max_ko if max_ko > 0 else 0,
        fighter1['submissions'] / max_sub if max_sub > 0 else 0,
        fighter1['win_streak'] / max_streak if max_streak > 0 else 0
    ]
    
    fighter2_values = [
        fighter2['striking_accuracy'],
        fighter2['takedown_accuracy'],
        fighter2['knockouts'] / max_ko if max_ko > 0 else 0,
        fighter2['submissions'] / max_sub if max_sub > 0 else 0,
        fighter2['win_streak'] / max_streak if max_streak > 0 else 0
    ]
    
    fig = go.Figure()
    
    # Add fighter 1 trace
    fig.add_trace(go.Bar(
        y=categories,
        x=fighter1_values,
        name=fighter1_name,
        orientation='h',
        marker_color='red',
        text=[f"{v:.2f}" for v in fighter1_values],
        textposition='auto'
    ))
    
    # Add fighter 2 trace with negative values for comparison
# Add fighter 2 trace with negative values for comparison
    fig.add_trace(go.Bar(
        y=categories,
        x=[-v for v in fighter2_values],  # Use negative values for side-by-side comparison
        name=fighter2_name,
        orientation='h',
        marker_color='blue',
        text=[f"{v:.2f}" for v in fighter2_values],  # Still show positive values in text
        textposition='auto'
    ))
    
    fig.update_layout(
        title='Fighter Comparison',
        xaxis=dict(
            title='',
            tickvals=[-1, -0.75, -0.5, -0.25, 0, 0.25, 0.5, 0.75, 1],
            ticktext=['1.0', '0.75', '0.5', '0.25', '0', '0.25', '0.5', '0.75', '1.0'],
            zeroline=True,
            zerolinewidth=2,
            zerolinecolor='white'
        ),
        yaxis=dict(
            title=''
        ),
        barmode='overlay',
        bargap=0.1,
        margin=dict(l=40, r=40, t=80, b=40),
        paper_bgcolor='rgba(0, 0, 0, 0)',
        plot_bgcolor='rgba(0, 0, 0, 0)',
        font=dict(color='white'),
        legend=dict(
            orientation="h",
            yanchor="bottom",
            y=1.02,
            xanchor="center",
            x=0.5
        )
    )
    
    return fig

# Head-to-head content callback
@app.callback(
    Output("head-to-head-content", "children"),
    [Input("fighter1-dropdown", "value"),
     Input("fighter2-dropdown", "value")]
)
def update_head_to_head(fighter1_name, fighter2_name):
    if not fighter1_name or not fighter2_name:
        return html.Div("Please select two fighters")
    
    # Get all matches between the two fighters
    head_to_head = matches_data[
        ((matches_data['fighter1'] == fighter1_name) & (matches_data['fighter2'] == fighter2_name)) |
        ((matches_data['fighter1'] == fighter2_name) & (matches_data['fighter2'] == fighter1_name))
    ].sort_values('date', ascending=False)
    
    if head_to_head.empty:
        return html.Div([
            html.H4("No previous fights", className="text-center text-muted my-4"),
            html.P("These fighters have not faced each other.", className="text-center")
        ])
    
    # Calculate the record between fighters
    fighter1_wins = head_to_head[head_to_head['winner'] == fighter1_name].shape[0]
    fighter2_wins = head_to_head[head_to_head['winner'] == fighter2_name].shape[0]
    
    return html.Div([
        html.H4("HEAD-TO-HEAD RECORD", className="text-center mb-3"),
        
        html.Div(className="d-flex justify-content-between mb-4", children=[
            html.Div(className="text-center", children=[
                html.H2(fighter1_wins, className="text-danger m-0"),
                html.P(fighter1_name, className="m-0")
            ]),
            
            html.Div(className="text-center", children=[
                html.H2("-", className="text-light m-0"),
            ]),
            
            html.Div(className="text-center", children=[
                html.H2(fighter2_wins, className="text-primary m-0"),
                html.P(fighter2_name, className="m-0")
            ]),
        ]),
        
        html.H5("PREVIOUS FIGHTS", className="text-center mb-2"),
        
        html.Div([
            dbc.Table([
                html.Tbody([
                    html.Tr([
                        html.Td([
                            html.Div(row['event']),
                            html.Small(row['date'], className="text-muted")
                        ]),
                        html.Td([
                            html.Div(row['winner'], className="text-success"),
                            html.Small(f"{row['method']} (R{row['round']})")
                        ])
                    ]) for _, row in head_to_head.iterrows()
                ])
            ], borderless=True, className="mb-0")
        ])
    ])

# Performance history callback
@app.callback(
    Output("performance-history", "figure"),
    [Input("fighter1-dropdown", "value"),
     Input("fighter2-dropdown", "value")]
)
def update_performance_history(fighter1_name, fighter2_name):
    if not fighter1_name or not fighter2_name:
        return {}
    
    # In a real app, this would use actual fight history data
    # For this example, we'll generate sample win/loss streaks
    
    # Sample fight dates
    dates = pd.date_range(start='2018-01-01', end='2024-01-01', periods=12)
    
    # Sample outcomes for fighter 1 (1 for win, 0 for loss)
    fighter1_outcomes = np.random.choice([0, 1], size=len(dates), p=[0.3, 0.7])
    # Cumulative wins for fighter 1
    fighter1_wins = np.cumsum(fighter1_outcomes)
    
    # Sample outcomes for fighter 2
    fighter2_outcomes = np.random.choice([0, 1], size=len(dates), p=[0.4, 0.6])
    # Cumulative wins for fighter 2
    fighter2_wins = np.cumsum(fighter2_outcomes)
    
    fig = go.Figure()
    
    fig.add_trace(go.Scatter(
        x=dates,
        y=fighter1_wins,
        mode='lines+markers',
        name=fighter1_name,
        line=dict(color='red', width=2),
        marker=dict(size=8)
    ))
    
    fig.add_trace(go.Scatter(
        x=dates,
        y=fighter2_wins,
        mode='lines+markers',
        name=fighter2_name,
        line=dict(color='blue', width=2),
        marker=dict(size=8)
    ))
    
    fig.update_layout(
        title='Career Wins Over Time',
        xaxis_title='',
        yaxis_title='Total Wins',
        legend_title='Fighter',
        margin=dict(l=40, r=40, t=40, b=40),
        paper_bgcolor='rgba(0, 0, 0, 0)',
        plot_bgcolor='rgba(0, 0, 0, 0)',
        font=dict(color='white')
    )
    
    return fig

# Run the app
if __name__ == '__main__':
    app.run_server(debug=True)
```

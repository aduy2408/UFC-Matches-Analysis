### df fighters

```python
df_fighters = pd.read_csv('fighters.csv')
```

```python
df_fighters.iloc[:5,  5:]
```

```python
df_fighters[['Wins', 'Losses', 'Draws']] = df_fighters['Record'].str.extract(r'(\d+)-(\d+)-(\d+)').astype(float)
df_fighters = df_fighters.drop(columns=['Record'])
```

```python
percentage_columns = ['Striking_Accuracy', 'Takedown_Accuracy','Sig_Str_Def','Takedown_Def']
for col in percentage_columns:
    df_fighters[col] = df_fighters[col].str.rstrip('%').astype(float)/100
```

```python
strike_columns = [
    'Sig_Str_Landed Total', 'Sig_Strikes_While_Clinched','Sig_Strikes_While_Standing',
    'Sig_Strikes_While_Grounded', 'Sig_Strikes_Head',
    'Sig_Strikes_Body', 'Sig_Strikes_Leg',
    'Win_by_KO/TKO', 'Win_by_Decision', 'Win_by_Submission'
]
for col in strike_columns:
    if df_fighters[col].dtype == 'object':
        df_fighters[[col + '_Total', col + '_Percent']] = df_fighters[col].str.extract(r'(\d+)\((\d+)%\)').astype(float)
        df_fighters[col + '_Percent'] = df_fighters[col + '_Percent'].fillna(0) 
        df_fighters[col + '_Percent'] = df_fighters[col + '_Percent'] / 100
    else:
        df_fighters[col + '_Total'] = df_fighters[col] 
        df_fighters[col + '_Percent'] = 0  
```

```python
df_fighters = df_fighters.drop(columns=[col for col in strike_columns if col in df_fighters.columns])
```

```python
def cv_time(time):
    if isinstance(time, str):
        parts = time.split(':')
        if len(parts) == 2:
            minutes, seconds = parts
            if minutes.isdigit() and seconds.isdigit():
                return int(minutes) * 60 + int(seconds)
    return 0
df_fighters['Avg_Fight_Time'] = df_fighters['Avg_Fight_Time'].apply(cv_time)
```

```python
df_fighters[df_fighters['Sig_Str_Def'].isna()].iloc[:10,8:]
```

```python
df_fighters['Takedowns_Landed_Total'] = df_fighters['Takedowns_Landed_Total'].fillna('0').astype(float)
```

```python
df_fighters['Weight_Class'] = df_fighters['Weight_Class'].fillna('0')
```

```python
df_fighters['Striking_Accuracy'] = df_fighters['Striking_Accuracy'].fillna('0').astype(float)
df_fighters['Takedown_Accuracy'] = df_fighters['Takedown_Accuracy'].fillna('0').astype(float)
df_fighters['Sig_Strikes_Per Min'] = df_fighters['Sig_Strikes_Per Min'].fillna('0').astype(float)
df_fighters['Takedown_Avg_Per Min'] = df_fighters['Takedown_Avg_Per Min'].fillna('0').astype(float)
df_fighters['Takedown_Avg_Per Min'] = df_fighters['Takedown_Avg_Per Min'].fillna('0').astype(float)
df_fighters['Sig_Str_Def'] = df_fighters['Sig_Str_Def'].fillna('0').astype(float)
```

```python
columns_to_fill = [
    'Knockdown_Avg',
    'Sig_Strikes_Absorbed_Per_Min',
    'Sub_Avg_Per_Min',
    'Takedown_Def',
    'Wins',
    'Losses',
    'Draws',
    'Sig_Strikes_While_Clinched_Total',
    'Sig_Strikes_While_Standing_Total',
    'Sig_Strikes_While_Grounded_Total',
    'Sig_Strikes_Head_Total',
    'Sig_Strikes_Body_Total',
    'Sig_Strikes_Leg_Total',
]
```

```python
for col in columns_to_fill:
    df_fighters[col] = df_fighters[col].fillna('0').astype(float)
```
